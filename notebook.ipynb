## ***FastAPI***

- FastAPI is a high-performance open-source web framework used for **building backend APIs with Python**. Released in 2018 by Sebastián Ramírez, it is specifically **designed to eliminate boilerplate code and maximize development speed** while maintaining execution performance comparable to NodeJS and Go

- **async REST APIs** (an architectural pattern designed to handle long-running background tasks, such as video processing, large file generation, or third-party checkouts, without blocking the client or forcing the server connection to time out)

## ***GET and POST routes***

```python
from fastapi import FastAPI

app = FastAPI()

items = []

@app.get("/") 
def root(): 
    return {"Hello": "World"}

@app.post("/items")
def create_item(item: str):
    items.append(item)
    return item # or we can make it return the whole "items" list as well

@app.get("/items/{item_id}")
def get_item(item_id: int) -> str:
    item = items[item_id]
    return item
```

#### **Then on a NEW termimal, do**

`curl -X POST -H "Content-Type: application/json" 'http://127.0.0.1:8000/items?item=apple'`

- apple is appended

`curl -X POST -H "Content-Type: application/json" 'http://127.0.0.1:8000/items?item=pineapple'`

- pineapple is appended



#### **Then use GET**

`curl -X GET http://127.0.0.1:8000/items/0`

- returns apple

`curl -X GET http://127.0.0.1:8000/items/1`

- returns pineapple
